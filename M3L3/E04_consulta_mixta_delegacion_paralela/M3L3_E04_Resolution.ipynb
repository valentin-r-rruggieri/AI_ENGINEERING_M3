## Sección 1 — Consultas mixtas

Una query puede pedir dos cosas: “vacaciones y VPN”. En ese caso no hay un único agente correcto.

```text
query mixta -> lista de intents -> varios agentes -> merge ordenado
```


Reutilizamos el registro de agentes, pero el clasificador ahora devuelve una lista completa.


````python
knowledge_base = {
    "hr": [
        "Vacaciones: cada empleado tiene 15 días hábiles por año.",
        "Beneficios: el seguro médico inicia el primer día de trabajo.",
        "Licencias: registrar el pedido en PeopleOps y avisar al manager.",
    ],
    "tech": [
        "VPN: reiniciar el cliente, validar MFA y abrir ticket si persiste.",
        "Contraseña: restablecer desde el portal de identidad.",
        "Notebook: reportar equipo dañado con número de serie.",
    ],
    "billing": [
        "Facturas: cargar comprobantes antes del día 25.",
        "Reembolsos: adjuntar recibo, monto y centro de costo.",
        "Pagos: se procesan los viernes por la tarde.",
    ],
}

KEYWORDS = {
    "hr": ["vacaciones", "beneficio", "seguro", "licencia"],
    "tech": ["vpn", "contraseña", "mfa", "notebook"],
    "billing": ["factura", "facturas", "reembolso", "pago", "recibo"],
}

def detect_domains(query: str) -> list[str]:
    text = query.lower()
    matches = [domain for domain, words in KEYWORDS.items() if any(word in text for word in words)]
    return matches or ["unknown"]

def retrieve(domain: str, query: str, k: int = 2) -> list[str]:
    docs = knowledge_base[domain]
    query_words = set(query.lower().replace("¿", "").replace("?", "").split())
    def score(doc: str) -> int:
        return sum(1 for word in query_words if word.strip(",.") in doc.lower())
    return sorted(docs, key=score, reverse=True)[:k]

def make_agent(domain: str, name: str) -> Callable[[str], str]:
    return lambda query: f"{name}: " + " | ".join(retrieve(domain, query, 1))
AGENTS = {"hr": make_agent("hr", "HRAgent"), "tech": make_agent("tech", "TechAgent"), "billing": make_agent("billing", "BillingAgent")}
````

## Sección 2 — Detectar, ejecutar y combinar

`detect_multiple_intents` devuelve `list[str]`; `run_agents` ejecuta cada especialista; `merge_responses` arma una salida legible.


````python
def detect_multiple_intents(query: str) -> list[str]:
    return detect_domains(query)

def run_agents(intents: list[str], query: str) -> list[dict]:
    return [{"intent": intent, "answer": AGENTS[intent](query)} for intent in intents if intent in AGENTS]

def merge_responses(responses: list[dict]) -> str:
    if not responses:
        return "No pude identificar un dominio claro."
    return "Respuesta combinada:\n" + "\n".join(f"- {r['intent'].upper()}: {r['answer']}" for r in responses)

def handle_query(query: str) -> str:
    return merge_responses(run_agents(detect_multiple_intents(query), query))
````

## Sección 3 — Casos representativos

Probamos consultas de dos dominios, tres dominios y fuera de alcance.


````python
for q in ["vacaciones y vpn", "reembolso y contraseña", "seguro factura notebook", "almuerzo"]:
    print("\nQUERY", q)
    print(handle_query(q))
````
