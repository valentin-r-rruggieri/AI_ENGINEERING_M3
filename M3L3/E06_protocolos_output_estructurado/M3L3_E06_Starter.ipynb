## Sección 1 — Por qué protocolizar

Si cada agente devuelve texto libre, el orquestador adivina. Con `AgentResponse`, lee `status` y decide.

| Status | Acción |
|---|---|
| `success` | mostrar respuesta |
| `needs_clarification` | pedir dato faltante |
| `out_of_scope` | fallback o reroute |


Creamos el contrato `AgentResponse` con campos estándar.


```python
class AgentResponse(TypedDict):
    agent_name: str
    status: Literal["success", "needs_clarification", "out_of_scope"]
    answer: str
    confidence: Literal["high", "medium", "low"]
    needs_handoff: bool
    target_agent: str | None
    sources: list[str]
```

## Sección 2 — Agentes y renderer

Los agentes devuelven protocolo. `render_response` convierte estado estructurado en texto para el usuario.

## Sección 3 — Probar estados
