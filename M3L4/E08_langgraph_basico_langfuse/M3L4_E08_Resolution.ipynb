## ¿Por qué importa este ejercicio?

Hasta ahora construimos tracing manual con MiniTracer. Pero en producción queremos algo más robusto: **LangGraph** para orquestar agentes y **Langfuse** para tracing automático.

Langfuse se integra con LangGraph mediante un **CallbackHandler** que captura cada paso del grafo y lo envía a la plataforma de observabilidad sin que tengas que instrumentar manualmente.

| Concepto | Definición simple | Cómo aparece acá |
|---|---|---|
| **LangGraph** | Framework para orquestar agentes como grafos | `StateGraph`, `add_node`, `compile()` |
| **Langfuse** | Plataforma de observabilidad para LLMs | `CallbackHandler()` que captura traces |
| **CallbackHandler** | Hook que Langfuse inyecta en LangGraph para tracear | `config={'callbacks': [langfuse_handler]}` |
| **State** | Estado compartido que fluye por los nodos del grafo | `TypedDict` con `messages`

```python
from typing import Annotated
from typing_extensions import TypedDict
from langgraph.graph import StateGraph, START, END
from langgraph.graph.message import add_messages
from langchain_core.messages import HumanMessage
from langchain_openai import ChatOpenAI
from langfuse.langchain import CallbackHandler

class State(TypedDict):
    messages: Annotated[list, add_messages]

llm = ChatOpenAI(model='gpt-4o-mini', temperature=0)
print('Imports OK.')
```

## Solución — Grafo básico

Definimos un grafo con un solo nodo `chatbot` que recibe mensajes y responde usando el LLM.

> **¿Por qué `State` usa `messages`?** Porque LangGraph maneja el historial de conversación como parte del estado. `add_messages` es un reducer que acumula mensajes.

```python
def chatbot_node(state: State) -> dict:
    response = llm.invoke(state['messages'])
    return {'messages': [response]}

builder = StateGraph(State)
builder.add_node('chatbot', chatbot_node)
builder.set_entry_point('chatbot')
builder.set_finish_point('chatbot')
graph = builder.compile()

print('Grafo compilado.')
print(graph.get_graph().draw_mermaid())
```

## Ejecución con Langfuse

Invocamos el grafo con el `CallbackHandler` de Langfuse. Cada ejecución generará un trace automáticamente en Langfuse.

```python
langfuse_handler = CallbackHandler()

result = graph.invoke(
    {'messages': [HumanMessage(content='Explicame qué es tracing en IA en 2 oraciones')]},
    config={'callbacks': [langfuse_handler]}
)

print('Respuesta:', result['messages'][-1].content)
```
