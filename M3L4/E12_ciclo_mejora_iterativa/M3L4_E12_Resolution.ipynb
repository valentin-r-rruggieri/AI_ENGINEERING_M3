## ¿Por qué importa este ejercicio?

Este es el ejercicio que cierra M3L4. Integra **todo** lo aprendido:

- Comparación de versiones (E05)
- Evaluación de calidad (E06)

- **Ciclo de mejora**: medir -> diagnosticar -> fixear -> re-medir -> documentar

El objetivo es simular un **fix real**: detectamos que el router falla en consultas de finance, diagnosticamos la causa, implementamos una versión mejorada, y documentamos el cambio.

| Concepto | Definición simple | Cómo aparece acá |
|---|---|---|
| **Ciclo iterativo** | Medir -> Diagnosticar -> Fixear -> Re-medir -> Documentar | Flujo completo del notebook |
| **Fix report** | Documentación del cambio y su impacto | Reporte markdown generado al final |
| **Before/After** | Comparación de métricas antes y después | `acc_v1` vs `acc_v2` |
| **Causa raíz** | Por qué ocurre el problema | Keywords insuficientes en router v1 |

```python
!pip install pandas -q
import pandas as pd

def route_query_v1(query):
    q = query.lower()
    if 'vacaciones' in q: return 'hr'
    if 'vpn' in q or 'error' in q: return 'it'
    if 'factura' in q: return 'finance'
    if 'contrato' in q: return 'legal'
    return 'general'

def route_query_v2(query):
    q = query.lower()
    hr_kw      = ['vacaciones','licencia','recibo','nómina','beneficios','rrhh']
    it_kw      = ['vpn','error','app','laptop','wifi','login','contraseña','sistema','acceso']
    finance_kw = ['factura','pago','reembolso','gasto','cobro','comprobante','salario']
    legal_kw   = ['contrato','legal','confidencialidad','nda','acuerdo','firma']
    det = []
    if any(w in q for w in hr_kw): det.append('hr')
    if any(w in q for w in it_kw): det.append('it')
    if any(w in q for w in finance_kw): det.append('finance')
    if any(w in q for w in legal_kw): det.append('legal')
    if len(det) > 1: return 'multi_intent'
    if len(det) == 1: return det[0]
    if len(q.split()) <= 2: return 'clarification'
    return 'general'

golden_dataset = [
    {'id': 'case_001', 'query': '¿Cómo solicito mis días de vacaciones?',          'expected_intent': 'hr'},
    {'id': 'case_002', 'query': 'Mi VPN no conecta desde ayer',                     'expected_intent': 'it'},
    {'id': 'case_003', 'query': 'Necesito ver mi factura del mes pasado',            'expected_intent': 'finance'},
    {'id': 'case_004', 'query': 'Necesito el contrato de confidencialidad actualizado', 'expected_intent': 'legal'},
    {'id': 'case_005', 'query': 'No puedo entrar al portal para ver mi recibo',     'expected_intent': 'hr'},
    {'id': 'case_006', 'query': 'ayuda',                                             'expected_intent': 'clarification'},
    {'id': 'case_007', 'query': '¿Cuándo se procesa el reembolso de gastos?',       'expected_intent': 'finance'},
    {'id': 'case_008', 'query': 'El sistema de login no me deja entrar',             'expected_intent': 'it'},
]
print('Setup listo.')
```

```python
def evaluate_router(router_fn, dataset):
    rows = []
    for case in dataset:
        actual = router_fn(case['query'])
        rows.append({'id': case['id'], 'query': case['query'],
                     'expected_intent': case['expected_intent'],
                     'actual_intent': actual,
                     'correct': int(actual == case['expected_intent'])})
    df = pd.DataFrame(rows)
    return df, df['correct'].mean()

df_v1, acc_v1 = evaluate_router(route_query_v1, golden_dataset)
df_v2, acc_v2 = evaluate_router(route_query_v2, golden_dataset)
print(f'Router v1: {acc_v1:.2%}')
print(f'Router v2: {acc_v2:.2%}')
print(f'Mejora: +{(acc_v2-acc_v1):.2%}')
```

## Paso 1 — Diagnosticar fallos de v1

Veamos exactamente qué casos fallaban en la versión original.

```python
print('Fallos en v1:')
df_v1[df_v1['correct']==0][['id','query','expected_intent','actual_intent']]
```

## Paso 2 — Comparar resultados

```python
comparison = pd.DataFrame([
    {'version': 'router_v1', 'routing_accuracy': acc_v1},
    {'version': 'router_v2', 'routing_accuracy': acc_v2}
])
comparison
```

## Paso 3 — Generar reporte de fix

Documentamos el análisis, la causa raíz, el cambio aplicado y las métricas. Este reporte es el **entregable** del ciclo de mejora.

```python
report = f"""
# Fix Report — Router Finance/HR Misclassification

## Issue
Varias consultas de finanzas y RRHH estaban siendo ruteadas a GeneralAgent,
resultando en respuestas genéricas e irrelevantes para el usuario.

## Causa raíz
El router v1 solo reconocía la palabra 'factura' para finance,
pero no contemplaba términos de uso frecuente como:
- pago, reembolso, gasto, comprobante, cobro, salario (finance)
- licencia, recibo, nómina, beneficios (hr)
- login, contraseña, sistema, acceso (it)

Tampoco detectaba multi-intent ni consultas ambiguas de 1-2 palabras (clarification).

## Cambio aplicado
Se implementó router_v2 con:
- Listas expandidas de keywords por dominio
- Detección de multi_intent cuando hay keywords de 2+ dominios
- Detección de clarification para queries muy cortas (<= 2 palabras)

## Métricas
| Versión | Routing Accuracy |
|---|---:|
| router_v1 | {acc_v1:.2%} |
| router_v2 | {acc_v2:.2%} |
| Mejora | +{(acc_v2-acc_v1):.2%} |

## Validación
El cambio fue validado usando golden_dataset_v1 ({len(golden_dataset)} casos).
Todos los casos de finance ahora se enrutan correctamente.

## Riesgo pendiente
- Consultas ambiguas entre dominios (ej: 'error en mi factura' -> it o finance?)
- El router sigue siendo basado en keywords — no entiende semántica
- Próximo paso: evaluar con un router LLM para mejor precisión en edge cases
"""
print(report)
```

## Verificación

```python
assert acc_v2 >= acc_v1
df_v2_fin, _ = evaluate_router(route_query_v2, [c for c in golden_dataset if c['expected_intent'] == 'finance'])
finance_acc_v2 = df_v2_fin['correct'].mean()
assert finance_acc_v2 > 0
print(f'Checks E12 OK')
print(f'v1: {acc_v1:.2%} -> v2: {acc_v2:.2%} (+{(acc_v2-acc_v1):.2%})')
print(f'Finance accuracy v2: {finance_acc_v2:.2%}')
```

## [OK] Cierre — El ciclo completo de mejora

| Paso | Qué hicimos | Herramienta |
|---|---|---|
| **1. Medir** | Ejecutamos golden dataset contra router v1 | `evaluate_router()` |
| **2. Diagnosticar** | Identificamos keywords faltantes | Análisis de fallos por dominio |
| **3. Fixear** | Implementamos router v2 con más keywords | `route_query_v2()` |
| **4. Re-medir** | Ejecutamos golden dataset contra router v2 | `evaluate_router()` |
| **5. Documentar** | Generamos fix report con métricas | Reporte markdown |

> **Esto es ingeniería de IA en producción:** no se trata de escribir código que funciona, sino de **medir, mejorar y documentar** cada cambio de forma sistemática.
