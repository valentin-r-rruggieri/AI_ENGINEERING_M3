## ¿Por qué importa este ejercicio?

En E01 construimos el MiniTracer como clase aislada. Pero la tracería **solo es útil si está integrada en el sistema que queremos observar**.

En este ejercicio vamos a **instrumentar** un sistema multiagente: cada request que entra genera un trace automáticamente, con spans para cada paso (routing, agente especialista).

| Concepto | Definición simple | Cómo aparece acá |
|---|---|---|
| **Instrumentación** | Agregar código de tracing dentro del sistema | `handle_query_with_tracing()` crea spans automáticamente |
| **Router** | Función que clasifica la consulta en un dominio | `route_query()` con keywords por dominio |
| **Agente especialista** | Función que responde según el dominio | `hr_agent()`, `it_agent()`, `finance_agent()`, `legal_agent()` |
| **Trace completo** | Trace con routing + agente en un flujo | `trace` con 2 spans (routing + agente) |

```python
import time
import uuid
from datetime import datetime

class MiniTracer:
    def __init__(self):
        self.traces = []

    def start_trace(self, name, input_data=None, metadata=None, tags=None):
        trace = {
            'trace_id': str(uuid.uuid4()),
            'name': name,
            'input': input_data,
            'output': None,
            'metadata': metadata or {},
            'tags': tags or [],
            'spans': [],
            'created_at': datetime.utcnow().isoformat(),
            '_started_at_ts': time.time()
        }
        self.traces.append(trace)
        return trace

    def add_span(self, trace, name, input_data=None, output_data=None, metadata=None, duration_ms=None):
        span = {
            'span_id': str(uuid.uuid4()),
            'name': name,
            'input': input_data,
            'output': output_data,
            'metadata': metadata or {},
            'duration_ms': duration_ms
        }
        trace['spans'].append(span)
        return span

    def update_trace_output(self, trace, output_data):
        trace['output'] = output_data
        trace['total_duration_ms'] = round((time.time() - trace['_started_at_ts']) * 1000, 2)

    def show_trace(self, trace):
        return trace

print('MiniTracer listo.')
```

```python
def route_query(query: str) -> str:
    q = query.lower()
    if any(w in q for w in ['vacaciones', 'licencia', 'recibo', 'nómina', 'portal rrhh']):
        return 'hr'
    if any(w in q for w in ['vpn', 'laptop', 'error', 'app', 'wifi', 'login', 'contraseña']):
        return 'it'
    if any(w in q for w in ['factura', 'pago', 'reembolso', 'gasto', 'cobro']):
        return 'finance'
    if any(w in q for w in ['contrato', 'legal', 'confidencialidad', 'nda']):
        return 'legal'
    if len(q.split()) <= 2:
        return 'clarification'
    return 'general'

def hr_agent(query): return 'Soy HRAgent. Te ayudo con vacaciones, licencias, recibos y temas de RRHH.'
def it_agent(query): return 'Soy ITAgent. Te ayudo con VPN, laptop, accesos y errores técnicos.'
def finance_agent(query): return 'Soy FinanceAgent. Te ayudo con facturas, pagos y reembolsos.'
def legal_agent(query): return 'Soy LegalAgent. Te ayudo con contratos y temas legales.'
def general_agent(query): return 'Necesito más contexto para ayudarte correctamente.'

agents = {'hr': hr_agent, 'it': it_agent, 'finance': finance_agent,
          'legal': legal_agent, 'general': general_agent, 'clarification': general_agent}

print('Agentes listos.')
```

## Solución — Función instrumentada

La función `handle_query_with_tracing()` integra el MiniTracer dentro del flujo del sistema:

1. **Inicia un trace** con la consulta del usuario como input
2. **Mide el tiempo** del routing y crea un span con `duration_ms` real
3. **Ejecuta el agente** correspondiente y crea otro span
4. **Actualiza el output** del trace con el resultado final

> **¿Por qué medimos `duration_ms` con `time.time()`?** Para simular tracing real donde cada paso tiene latencia. En producción estos valores saldrían de medir llamadas a APIs o bases de datos.

```python
def handle_query_with_tracing(query: str, tracer: MiniTracer) -> dict:
    trace = tracer.start_trace(
        name='multiagent-support-request',
        input_data={'query': query},
        metadata={'environment': 'notebook', 'router_version': 'v1'},
        tags=['multiagent', 'm3l4']
    )

    start = time.time()
    intent = route_query(query)
    duration_routing = round((time.time() - start) * 1000, 2)

    tracer.add_span(
        trace,
        name='orchestrator-routing',
        input_data={'query': query},
        output_data={'intent': intent},
        metadata={'router_version': 'v1'},
        duration_ms=duration_routing
    )

    selected_agent = agents.get(intent, general_agent)

    start = time.time()
    response = selected_agent(query)
    duration_agent = round((time.time() - start) * 1000, 2)

    tracer.add_span(
        trace,
        name=f'{intent}-agent',
        input_data={'query': query},
        output_data={'response': response},
        duration_ms=duration_agent
    )

    tracer.update_trace_output(trace, {'intent': intent, 'final_response': response})
    return trace

print('Función lista.')
```

## Ejecución — Una consulta

Probamos con una consulta de factura. El trace resultante muestra los 2 spans con sus duraciones reales.

```python
tracer = MiniTracer()
trace = handle_query_with_tracing('No puedo ver mi factura', tracer)
trace
```

## Ejecución — Múltiples consultas

Probamos todo el espectro de dominios. Cada consulta genera su propio trace con los spans correspondientes.

```python
queries = [
    '¿Cómo solicito mis días de vacaciones?',
    'Mi VPN no conecta desde ayer',
    'Necesito ver mi factura del mes pasado',
    'Necesito el contrato de confidencialidad actualizado',
    'ayuda'
]

tracer2 = MiniTracer()
for q in queries:
    t = handle_query_with_tracing(q, tracer2)
    print(f"Query: {q[:45]}")
    print(f"  Intent: {t['output']['intent']}")
    print(f"  Spans: {[s['name'] for s in t['spans']]}")
    print()
```

## Verificación

Validamos que el tracing funciona correctamente:
- El trace no es `None`
- Tiene exactamente 2 spans (routing + agente)
- El primer span es el routing
- El intent detectado es `finance`
- Ambos spans tienen duración registrada

```python
tracer_check = MiniTracer()
t1 = handle_query_with_tracing('No puedo ver mi factura', tracer_check)
assert t1 is not None
assert len(t1['spans']) == 2
assert t1['spans'][0]['name'] == 'orchestrator-routing'
assert t1['output']['intent'] == 'finance'
assert t1['spans'][0]['output']['intent'] == 'finance'
assert t1['spans'][1]['output']['response'] is not None
assert t1['spans'][0]['duration_ms'] is not None
assert t1['spans'][1]['duration_ms'] is not None
print('Checks E02 OK')
```
