## ¿Por qué importa este ejercicio?

En E04 armamos un golden dataset y medimos la precisión del router. Ahora vamos a dar el paso siguiente: **comparar dos versiones** del router y decidir cuál es mejor.

Esto es el corazón de la **ingeniería basada en datos**: no asumir que un cambio es mejora, sino **medirlo** contra un conjunto de referencia.

| Concepto | Definición simple | Cómo aparece acá |
|---|---|---|
| **Router v1** | Versión básica con keywords simples | Solo 1-2 keywords por dominio |
| **Router v2** | Versión mejorada con keywords expandidas | Listas completas + detección multi_intent + clarification |
| **A/B test** | Comparar dos versiones contra el mismo dataset | `evaluate_router(v1)` vs `evaluate_router(v2)` |
| **Mejora cuantificable** | Diferencia de accuracy entre versiones | `acc_v2 - acc_v1` |

```python
!pip install pandas -q
import pandas as pd

def route_query_v1(query):
    q = query.lower()
    if 'vacaciones' in q: return 'hr'
    if 'vpn' in q or 'error' in q: return 'it'
    if 'factura' in q: return 'finance'
    if 'contrato' in q: return 'legal'
    return 'general'

def route_query_v2(query):
    q = query.lower()
    hr_kw      = ['vacaciones', 'licencia', 'recibo', 'nómina', 'beneficios', 'rrhh', 'portal rrhh']
    it_kw      = ['vpn', 'error', 'app', 'laptop', 'wifi', 'login', 'contraseña', 'sistema', 'acceso']
    finance_kw = ['factura', 'pago', 'reembolso', 'gasto', 'cobro', 'comprobante', 'salario']
    legal_kw   = ['contrato', 'legal', 'confidencialidad', 'nda', 'acuerdo', 'firma']
    detected = []
    if any(w in q for w in hr_kw):      detected.append('hr')
    if any(w in q for w in it_kw):      detected.append('it')
    if any(w in q for w in finance_kw): detected.append('finance')
    if any(w in q for w in legal_kw):   detected.append('legal')
    if len(detected) > 1:   return 'multi_intent'
    if len(detected) == 1:  return detected[0]
    if len(q.split()) <= 2: return 'clarification'
    return 'general'

golden_dataset = [
    {'id': 'case_001', 'query': '¿Cómo solicito mis días de vacaciones?',             'expected_intent': 'hr'},
    {'id': 'case_002', 'query': 'Mi VPN no conecta desde ayer',                        'expected_intent': 'it'},
    {'id': 'case_003', 'query': 'Necesito ver mi factura del mes pasado',               'expected_intent': 'finance'},
    {'id': 'case_004', 'query': 'Necesito el contrato de confidencialidad actualizado', 'expected_intent': 'legal'},
    {'id': 'case_005', 'query': 'No puedo entrar al portal para ver mi recibo',        'expected_intent': 'hr'},
    {'id': 'case_006', 'query': 'ayuda',                                                'expected_intent': 'clarification'},
    {'id': 'case_007', 'query': '¿Cuándo se procesa el reembolso de gastos?',         'expected_intent': 'finance'},
    {'id': 'case_008', 'query': 'El sistema de login no me deja entrar',               'expected_intent': 'it'},
    {'id': 'case_009', 'query': 'Necesito cambiar mi contraseña',                       'expected_intent': 'it'},
    {'id': 'case_010', 'query': 'Mi laptop se apaga sola',                              'expected_intent': 'it'},
]
print('Setup listo.')
```

```python
def evaluate_router(router_fn, dataset):
    rows = []
    for case in dataset:
        actual = router_fn(case['query'])
        rows.append({
            'id': case['id'],
            'query': case['query'],
            'expected_intent': case['expected_intent'],
            'actual_intent': actual,
            'correct': int(actual == case['expected_intent'])
        })
    df = pd.DataFrame(rows)
    return df, df['correct'].mean()

print('Función lista.')
```

## Solución — Comparación v1 vs v2

Ejecutamos ambas versiones contra el mismo dataset y comparamos.

```python
df_v1, acc_v1 = evaluate_router(route_query_v1, golden_dataset)
df_v2, acc_v2 = evaluate_router(route_query_v2, golden_dataset)

print(f'Router v1 accuracy: {acc_v1:.2%}')
print(f'Router v2 accuracy: {acc_v2:.2%}')
print(f'Mejora: +{(acc_v2 - acc_v1):.2%}')
```

## Tabla comparativa

```python
comparison = pd.DataFrame([
    {'version': 'router_v1', 'routing_accuracy': acc_v1},
    {'version': 'router_v2', 'routing_accuracy': acc_v2}
])
comparison
```

## Casos donde difieren

Veamos exactamente qué casos cambian entre v1 y v2.

```python
merged = df_v1[['id','query','expected_intent','actual_intent']].merge(
    df_v2[['id','actual_intent']],
    on='id',
    suffixes=('_v1','_v2')
)
diff = merged[merged['actual_intent_v1'] != merged['actual_intent_v2']]
print(f'Casos donde v1 y v2 difieren: {len(diff)}')
diff
```

## Verificación

v2 debe ser >= v1 en accuracy.

```python
assert acc_v2 >= acc_v1
print(f'Checks E05 OK — v1: {acc_v1:.2%} -> v2: {acc_v2:.2%}')
```

## [OK] Cierre — ¿Qué aprendimos?

| Versión | Accuracy | Cambios clave |
|---|---|---|
| **Router v1** | 3 keywords simples | No detecta casos ambiguos ni consultas cortas |
| **Router v2** | Keywords expandidas + multi_intent + clarification | Cubre más variaciones del lenguaje natural |

> **Principio:** nunca deployes un cambio sin medir su impacto contra un golden dataset. Lo que no se mide, no se puede mejorar.
