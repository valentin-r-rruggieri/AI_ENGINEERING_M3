## ¿Por qué importa este ejercicio?

Cuando un agente en producción falla, **no podés mirar el código para entender qué pasó** — tenés que reconstruirlo a partir de los registros que dejó.

Los **logs tradicionales** (prints o archivos de texto) son frágiles: te dicen que algo pasó, pero no te dicen **qué se decidió, en qué orden, cuánto tardó o por qué falló**.

El **tracing estructurado** organiza la información en una jerarquía de pasos (spans) con input, output, duración y metadatos. Esto permite responder preguntas que los logs no pueden.

| Concepto | Definición simple | Cómo aparece en este notebook |
|---|---|---|
| **Log** | Evento suelto con un mensaje de texto | `print('INFO: request received')` |
| **Trace** | Historia completa de una request con estructura jerárquica | Diccionario con `trace_name`, `spans`, `output` |
| **Span** | Un paso atómico dentro del trace (routing, agente, retrieval) | Cada elemento en `trace['spans']` con `name`, `input`, `output`, `duration_ms` |
| **Timestamps** | Momento exacto de cada evento | `duration_ms` por span (en logs no hay) |
| **Contexto** | Input, output y metadata asociados a cada paso | `input`, `output`, `metadata` en cada span |

## Paso 1 — Logs tradicionales

Así se ve un sistema instrumentado solo con `print()`. Cada línea es un evento suelto, sin estructura, sin timestamps, sin relación entre sí.

```python
def process_request_with_logs(query: str):
    print('INFO: request received')
    print(f'INFO: user query = {query}')
    print('INFO: classified intent')
    print('INFO: agent executed')
    print('INFO: response returned')

process_request_with_logs('No puedo ver mi factura')
```

## Análisis: ¿Qué información falta en los logs?

Con solo prints, no podemos responder preguntas básicas de debugging.

| Pregunta | ¿Logs lo responden? | ¿Por qué? |
|---|---|---|
| ¿Qué **intent** se detectó? | [X] No | Solo dice 'classified intent', sin el valor |
| ¿Qué **agente** respondió? | [X] No | Solo dice 'agent executed' |
| ¿Cuánto **tardó** cada paso? | [X] No | No hay timestamps ni duraciones |
| ¿Hubo **retrieval vacío**? | [X] No | No hay outputs por paso |
| ¿Hubo **error** en un paso intermedio? | [X] No | Si no hay `ERROR:`, todo parece OK aunque haya fallado |

## Paso 2 — Trace estructurado

Ahora modelemos la misma request como un **trace**. Cada paso tiene nombre, input, output y duración. La estructura es jerárquica y consultable.

```python
trace = {
    'trace_name': 'support-request',
    'input': 'No puedo ver mi factura',
    'spans': [
        {
            'name': 'orchestrator-routing',
            'input': 'No puedo ver mi factura',
            'output': {'intent': 'finance'},
            'duration_ms': 120
        },
        {
            'name': 'finance-agent',
            'input': 'No puedo ver mi factura',
            'output': {'answer': 'Podés ver tu factura desde el portal de pagos.'},
            'duration_ms': 860
        }
    ],
    'output': {
        'final_answer': 'Podés ver tu factura desde el portal de pagos.'
    }
}

trace
```

## Revisión: El trace responde lo que los logs no podían

| Pregunta | Trace responde | Cómo |
|---|---|---|
| ¿Qué intent se detectó? | [OK] Sí | `spans[0].output.intent = 'finance'` |
| ¿Qué agente respondió? | [OK] Sí | `spans[1].name = 'finance-agent'` |
| ¿Cuánto tardó cada paso? | [OK] Sí | `spans[0].duration_ms = 120`, `spans[1].duration_ms = 860` |
| ¿Hubo retrieval vacío? | [OK] Sí | Se puede agregar span de retrieval con `output.count = 0` |
| ¿Dónde ocurrió el error? | [OK] Sí | El span con error queda marcado con contexto completo |

## Paso 3 — Trace con error (misclassification)

A veces el problema no es que el sistema se caiga, sino que **haga lo incorrecto**. Con un trace podemos detectar una **misclassification**: el router eligió un agente equivocado.

```python
bad_trace = {
    'trace_name': 'support-request',
    'input': {'query': 'No puedo ver mi factura'},
    'metadata': {
        'expected_intent': 'finance',
        'actual_intent': 'it'
    },
    'spans': [
        {
            'name': 'orchestrator-routing',
            'input': {'query': 'No puedo ver mi factura'},
            'output': {'intent': 'it'},
            'duration_ms': 140
        },
        {
            'name': 'it-agent',
            'input': {'query': 'No puedo ver mi factura'},
            'output': {'response': 'Probá reiniciar la app.'},
            'duration_ms': 900
        }
    ],
    'output': {
        'final_response': 'Probá reiniciar la app.'
    }
}

bad_trace
```

## Análisis del error — Resolución

1. **¿En qué span ocurrió el problema?** -> `orchestrator-routing`. Devolvió `it` en lugar de `finance`.
2. **Intent esperado vs real:** `finance` esperado, `it` detectado.
3. **¿Qué información del trace permite detectarlo?** -> `metadata.expected_intent` comparado con `spans[0].output.intent`.
4. **¿Qué cambio haría?** -> Mejorar las reglas del router: agregar 'factura', 'pago', 'comprobante' como keywords de `finance`.

> **Lección clave:** el trace no solo dice *qué* pasó, sino *dónde exactamente* en el flujo ocurrió la desviación. Con logs tradicionales esto es invisible.

## [OK] Cierre — Logs vs Traces: la diferencia fundamental

| Dimensión | Logs tradicionales | Tracing estructurado |
|---|---|---|
| **Estructura** | Eventos sueltos, sin contexto | Historia completa de la request |
| **Jerarquía** | Plana (líneas secuenciales) | Spans anidados en orden |
| **Input/Output** | No visible | Cada span tiene input y output |
| **Latencias** | Imposible sin timestamps manuales | `duration_ms` por span |
| **Misclassifications** | Invisibles | `metadata.expected_intent` vs `actual_intent` |
| **Debugging** | Buscar agujas en un pajar | Navegar la jerarquía de spans |

**Frase clave:** logs dicen *qué pasó*. Las trazas dicen *exactamente cómo pasó, por qué, y cuánto tardó cada parte*.
