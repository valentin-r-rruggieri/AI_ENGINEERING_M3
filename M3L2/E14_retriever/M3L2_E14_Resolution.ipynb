# M3L2 E14 - Retriever: la interfaz de búsqueda (Resolution)

## BLOQUE 1 — ¿Qué es un Retriever y para qué sirve?

### El problema

En E13 viste que `vectorstore.similarity_search(query, k)` busca documentos en FAISS. Pero:

- `similarity_search()` es específico de FAISS
- Si mañana cambiás a Chroma, Pinecone o Weaviate, cambiás la forma de buscar
- No podés enchufar el vector store directamente en una chain LCEL

### La solución

`vectorstore.as_retriever()` envuelve el vector store en una **interfaz estándar** llamada `BaseRetriever`.

```text
vectorstore (FAISS, Chroma, Pinecone, etc.)
  |
  |  .as_retriever(search_kwargs={"k": 2})
  v
retriever (BaseRetriever)
  |
  |  .invoke(query)  <- interfaz estandar
  v
List[Document]
```

El `retriever`:
- Usa la misma interfaz que `llm`, `prompt`, `parser` (todos responden a `.invoke()`)
- Se puede conectar en una chain LCEL con `|`
- Cambiar de vector store = cambiar 1 línea (la que crea el vectorstore)

## BLOQUE 2 — Retriever en el pipeline RAG

```text
Pregunta
  |
  v
Retriever (interfaz estandar)
  |  internamente: embed_query -> vectorstore.search
  v
Documentos relevantes
  |
  v
format_docs (funcion auxiliar)
  |  junta los textos en un solo string
  v
PromptTemplate + LLM + Parser
  |
  v
Respuesta
```

En una chain LCEL se ve así:

```python
chain = retriever | format_docs | prompt | llm | parser
```

Sin el retriever, tendrías que llamar `vectorstore.similarity_search()` a mano y pasar los resultados al prompt.

## BLOQUE 3 — Setup

````python
import os, getpass
if not os.getenv("OPENAI_API_KEY"):
    os.environ["OPENAI_API_KEY"] = getpass.getpass("OpenAI API key: ")

from langchain_openai import OpenAIEmbeddings
from langchain_community.vectorstores import FAISS

embeddings = OpenAIEmbeddings()
DOCS = [
    "La politica de vacaciones es de 15 dias por ano.",
    "El seguro medico esta incluido desde el primer dia.",
    "El horario de trabajo es de 9 a 18 con almuerzo.",
    "El trabajo remoto esta permitido 3 dias por semana.",
    "Los bonos anuales se pagan en diciembre.",
]
vectorstore = FAISS.from_texts(DOCS, embeddings)
print("Vector store listo.")
````

## BLOQUE 4 — Sin retriever: llamada directa al vector store

`similarity_search()` funciona pero es específica del vector store. Si cambiás a Chroma, el método puede ser distinto.

````python
# Sin retriever: metodo especifico de FAISS
docs_direct = vectorstore.similarity_search("vacaciones", k=2)
print("similarity_search (especifico de FAISS):")
for doc in docs_direct:
    print(f"  - {doc.page_content}")
print()
print("Si cambio a Chroma, .similarity_search() podria tener parametros distintos.")
````

## BLOQUE 5 — TODO 1: Crear el retriever con `as_retriever()`

```python
retriever = vectorstore.as_retriever(search_kwargs={"k": 2})
```

Esto devuelve un objeto `BaseRetriever` que:
- Se usa con `.invoke(query)` (igual que el LLM, el prompt, el parser)
- Es independiente del vector store subyacente
- Se puede conectar con `|` en una chain LCEL

````python
# TODO 1
retriever = vectorstore.as_retriever(search_kwargs={"k": 2})
print(f"Tipo del retriever: {type(retriever).__name__}")
print(f"Configuracion: {retriever.search_kwargs}")
print()
print("as_retriever() crea un objeto que responde a .invoke().")
````

## BLOQUE 6 — TODO 2: Invocar el retriever

`retriever.invoke(query)` devuelve una lista de objetos `Document` con el texto recuperado.

````python
# TODO 2
docs_retriever = retriever.invoke("vacaciones")
print("as_retriever + invoke:")
for doc in docs_retriever:
    print(f"  - {doc.page_content}")
print()
print("Mismo resultado que similarity_search.")
print("Pero si cambias FAISS por Chroma, solo cambia la linea del vectorstore.")
````

## BLOQUE 7 — TODO 3: Comparar distintos valores de k

El parámetro `k` controla el **tradeoff entre cantidad de contexto y costo**.

| k = 1 | k = 3 |
|---|---|
| Respuesta enfocada | Contexto más rico |
| Menos tokens = menor costo | Más tokens = mayor costo |
| Puede omitir info relevante | Puede incluir ruido |

Se pueden crear distintos retrievers para distintos casos de uso.

````python
# TODO 3
retriever_k1 = vectorstore.as_retriever(search_kwargs={"k": 1})
retriever_k3 = vectorstore.as_retriever(search_kwargs={"k": 3})
query = "politica de empresa"

print(f"Query: '{query}'")
print(f"  k=1: {len(retriever_k1.invoke(query))} documento(s)")
print(f"  k=3: {len(retriever_k3.invoke(query))} documento(s)")
print()
for i, doc in enumerate(retriever_k3.invoke(query)):
    print(f"  Doc {i+1}: {doc.page_content}")
print()
print("Tradeoff: k mayor = mas contexto pero mas tokens.")
````

## BLOQUE 8 — El retriever en una chain LCEL

El retriever se conecta directamente con `|` porque implementa la interfaz `Runnable`.

````python
from langchain_core.prompts import ChatPromptTemplate
from langchain_openai import ChatOpenAI
from langchain_core.output_parsers import StrOutputParser

def format_docs(docs):
    return "\n\n".join(d.page_content for d in docs)

prompt_rag = ChatPromptTemplate.from_messages([
    ("system", "Eres un asistente de RRHH. Responde solo con el contexto dado."),
    ("human", "Contexto:\n{context}\n\nPregunta:\n{question}"),
])

llm = ChatOpenAI(model="gpt-4o-mini", temperature=0)

# Chain RAG completa: retriever -> format -> prompt -> llm -> parser
rag_chain = (
    {"context": retriever | format_docs, "question": lambda x: x}
    | prompt_rag
    | llm
    | StrOutputParser()
)

respuesta = rag_chain.invoke("Politica de vacaciones?")
print(f"Respuesta RAG: {respuesta}")
````

## BLOQUE 9 — Errores comunes

| Error | Consecuencia | Solución |
|---|---|---|
| Llamar `retriever.invoke()` sin `search_kwargs` | Usa k=4 por defecto (puede no ser ideal) | Siempre especificá `search_kwargs={"k": n}` |
| Olvidar `format_docs` en la chain | El retriever devuelve `List[Document]`, no un string | Usá `retriever | format_docs` para convertir a texto |
| Usar `retriever` directamente como string en el prompt | Error de tipo: el prompt espera string, recibe lista | Asegurate de formatear los documentos primero |
| Cambiar el vector store sin cambiar el retriever | El retriever apunta al store viejo | Creá un nuevo retriever con `.as_retriever()` |
| No testear con distintos valores de k | Respuestas incompletas o con ruido | Probá k=1, k=3, k=5 y compará resultados |

## BLOQUE 10 — Comparación final

| Aspecto | Sin retriever (`similarity_search`) | Con retriever (`as_retriever`) |
|---|---|---|
| **API** | Específica del vector store (FAISS: `.similarity_search()`, Chroma: `.similarity_search_by_vector()`) | Estándar: `.invoke(query)` |
| **Componible con LCEL** | No (hay que llamarlo aparte) | Sí: `retriever | format_docs | prompt | llm | parser` |
| **Cambiar de store** | Cambiar todas las llamadas a `similarity_search` | Cambiar solo `vectorstore = FAISS(...)` → `vectorstore = Chroma(...)` |
| **Configuración** | Parámetros en cada llamada | `search_kwargs` en un solo lugar |
| **Testing** | Difícil de mockear | Fácil: se puede reemplazar el retriever |

### Regla práctica

```text
Si usas vector stores en LangChain:
  SIEMPRE usa retriever = vectorstore.as_retriever()
  NUNCA llames similarity_search() directamente
  
Motivo: el retriever es la interfaz estandar.
Si despues cambias de base de datos vectorial,
tu codigo no cambia.
```
