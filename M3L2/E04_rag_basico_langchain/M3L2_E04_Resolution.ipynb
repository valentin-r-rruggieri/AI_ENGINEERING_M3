## 1. ¿Qué es RAG y por qué existe?

**RAG** = Retrieval Augmented Generation

### El problema

Los LLMs (GPT, Claude, Gemini) tienen dos limitaciones graves:
1. **Cutoff de entrenamiento**: no saben nada después de su fecha de entrenamiento
2. **No conocen tus datos**: no saben tu política de RRHH, tu código interno, tus documentos privados
3. **Alucinaciones**: cuando no saben algo, lo inventan con total seguridad

### La solución

RAG divide el problema en 2 pasos:

```
1. RECUPERAR: buscar en tus documentos los fragmentos relevantes para la pregunta
2. GENERAR: pasarle al LLM solo esos fragmentos como contexto + la pregunta
```

El LLM **no necesita saber** toda tu base de conocimiento. Solo necesita ver los 2-3 fragmentos relevantes para responder esa pregunta específica.

### Flujo completo

```text
FASE UNICA (indexacion):
  Documentos -> Splitter -> Chunks -> Embeddings -> Vector Store

CADA PREGUNTA (consulta):
  Pregunta -> Embedding -> Buscar en Vector Store -> k chunks relevantes
                                                       |
                                                       v
  Prompt (contexto + pregunta) -> LLM -> Respuesta
```

### RAG vs otras alternativas

| Aspecto | Fine-tuning | RAG |
|---|---|---|
| Costo | $$$ (re-entrenar modelo) | $ (solo indexar) |
| Actualizar datos | Re-entrenar cada vez | Cambiar docs en el vector store |
| Transparencia | Caja negra | Sabés qué docs usó |
| Escala | El modelo aprende info general | Millones de documentos |
| Velocidad de implementación | Semanas | Días con LangChain |

## 4. Crear documento de ejemplo

Simulamos un archivo de texto con información de RRHH. En producción esto sería un PDF, una web, una base de datos, etc.

## 5. Loader: cargar el documento

**¿Qué hace?** Toma un archivo (o URL, PDF, etc.) y lo convierte en un objeto `Document` de LangChain.

**Tipos de loaders en LangChain:**

| Loader | Fuente | Instalación extra |
|---|---|---|
| `TextLoader` | Archivos .txt | Ninguna |
| `PyPDFLoader` | PDFs | `pypdf` |
| `CSVLoader` | Archivos CSV | Ninguna |
| `WebBaseLoader` | Páginas web | `beautifulsoup4` |
| `GitHubLoader` | Repos de GitHub | `pygithub` |
| `JSONLoader` | Archivos JSON | `jq` |
| `UnstructuredExcelLoader` | Excel/Word/PPT | `unstructured` |

El output del loader es: `List[Document]` donde cada `Document` tiene `page_content` (el texto) y `metadata` (fuente, página, etc.).

````python
loader = TextLoader("documento.txt", encoding="utf-8")
docs = loader.load()

print(f"Cantidad de documentos: {len(docs)}")
print(f"Tipo: {type(docs[0]).__name__}")
print(f"Contenido:\n{docs[0].page_content}")
print(f"Metadata: {docs[0].metadata}")
````

## 6. Splitter: dividir en fragmentos

**¿Por qué dividir?**
- Los LLMs tienen límite de tokens de contexto
- Si mandás el documento entero, el LLM recibe muchísimo texto y puede perderse
- Si dividís en fragmentos, el retriever puede buscar **solo el fragmento relevante**

**`RecursiveCharacterTextSplitter`**:
- Divide el texto recursivamente por separadores (`\n\n`, `\n`, `.`, espacio)
- Intenta mantener los chunks completos (no corta palabras)

### Parámetros clave

| Parámetro | Default | ¿Qué controla? |
|---|---|---|
| `chunk_size` | 1000 | Tamaño máximo de cada chunk en caracteres |
| `chunk_overlap` | 200 | Superposición entre chunks consecutivos |
| `separators` | `["\n\n", "\n", ".", " "]` | Por dónde intentar dividir (en orden) |

### ¿Cómo elegir chunk_size?

| Si tu documento... | Usá chunk_size... |
|---|---|
| Tiene párrafos cortos (políticas, FAQs) | 300-500 |
| Tiene párrafos medianos (artículos, manuales) | 500-1000 |
| Tiene secciones largas (libros, informes) | 1000-2000 |

### ¿Para qué sirve el overlap?

```text
Texto original:
"El modulo 3 trabaja con agentes, LangChain, RAG y trazabilidad."

Sin overlap:
  Chunk 1: "El modulo 3 trabaja con agentes, LangChain,"
  Chunk 2: "RAG y trazabilidad."
  -> Si buscas "LangChain", el chunk 2 no aparece aunque está relacionado

Con overlap=20:
  Chunk 1: "El modulo 3 trabaja con agentes, LangChain, RAG"
  Chunk 2: "LangChain, RAG y trazabilidad."
  -> Ambos chunks mencionan LangChain y RAG = mejor recuperación
```

````python
splitter = RecursiveCharacterTextSplitter(
    chunk_size=120,
    chunk_overlap=20,
)
chunks = splitter.split_documents(docs)

print(f"Cantidad de chunks: {len(chunks)}")
for i, chunk in enumerate(chunks):
    print(f"\n--- Chunk {i+1} ({len(chunk.page_content)} chars) ---")
    print(chunk.page_content)
````

---
## 7. Embeddings + Chroma: indexar vectores

### ¿Qué son los embeddings?

Un **embedding** es una lista de números (1536 para OpenAI) que representa el **significado** de un texto.

```text
"vacaciones" -> [0.012, -0.045, 0.089, ..., 0.098]  (1536 números)
"licencia"   -> [0.011, -0.042, 0.091, ..., 0.095]  (casi igual -> mismo significado)
"futbol"     -> [-0.023, 0.078, -0.012, ..., 0.034] (muy diferente)
```

Textos con significado similar tienen vectores **cercanos** (distancia pequeña). Textos diferentes tienen vectores **lejanos**.

### ¿Qué es Chroma?

Chroma es una **base de datos vectorial** open-source. Guarda:
- El vector (1536 números)
- El texto original
- Metadata (fuente, página, etc.)

Luego permite buscar: "dame los k vectores más cercanos a este otro vector"

### `Chroma.from_documents(chunks, embeddings)` hace:

```text
Por cada chunk:
  1. embeddings.embed_query(chunk) -> vector [1536 dims]
  2. Chroma guarda: vector + texto original + metadata
  3. Crea un indice para busqueda rapida
```

### `as_retriever(search_kwargs={"k": 2})`

Crea un **Retriever**: interfaz estándar para buscar documentos.

```python
retriever.invoke("pregunta")  # devuelve los k documentos mas relevantes
```

El parámetro **k** controla cuántos documentos recuperar:
- k=1: respuesta enfocada, mínimo costo de tokens
- k=3: contexto más rico
- k=5+: mucho contexto, pero puede incluir ruido

**Valor típico en producción: k=3 a k=5**

````python
embeddings = OpenAIEmbeddings()
vectorstore = Chroma.from_documents(chunks, embeddings)
retriever = vectorstore.as_retriever(search_kwargs={"k": 2})

print(f"Vector store: {type(vectorstore).__name__}")
print(f"Retriever: {type(retriever).__name__}")
print(f"Chunks en el indice: {vectorstore._collection.count()}")
````

---
## 8. Recuperar contexto con el retriever

Probamos el retriever por separado. Internamente hace:
1. `embeddings.embed_query(pregunta)` -> vector de la pregunta
2. Chroma busca los `k` vectores más cercanos a ese vector
3. Devuelve los `k` objetos `Document` con el texto original

Esto es el **corazón del RAG**: convertir la pregunta a vector, buscar en el índice, devolver solo los fragmentos relevantes.

````python
pregunta = "Que permite hacer RAG?"
fragmentos = retriever.invoke(pregunta)
contexto = "\n\n".join(d.page_content for d in fragmentos)

print(f"Pregunta: '{pregunta}'")
print(f"Fragmentos recuperados: {len(fragmentos)}")
for i, doc in enumerate(fragmentos):
    print(f"\n  [{i+1}] {doc.page_content}")
print(f"\nContexto formateado (entrada al prompt):\n{contexto}")
````

---
## 9. Prompt + LLM + Parser: generar la respuesta

### El prompt RAG

Tiene 2 partes clave:
1. **System message**: instrucción al modelo de que solo use el contexto dado
2. **Human message**: el contexto recuperado + la pregunta del usuario

```python
prompt = ChatPromptTemplate.from_messages([
    ("system", "Responde solo con el contexto..."),
    ("human", "Contexto:\n{contexto}\n\nPregunta: {pregunta}"),
])
```

### ¿Por qué el system message es tan importante?

Sin la instrucción "responde solo con el contexto", el LLM puede:
- Usar su conocimiento interno en vez del contexto
- Inventar información que no está en los documentos
- Ignorar el contexto cuando "cree saber" la respuesta

### La chain completa

```text
prompt | llm | parser
  ^       ^      ^
  |       |      +-- StrOutputParser: extrae .content del AIMessage
  |       +--------- ChatOpenAI: genera la respuesta
  +----------------- ChatPromptTemplate: arma los mensajes
```

````python
prompt = ChatPromptTemplate.from_messages([
    ("system", "Responde solo con el contexto proporcionado. "
                "Si la respuesta no esta en el contexto, "
                "di que no tienes informacion suficiente."),
    ("human", "Contexto:\n{contexto}\n\nPregunta: {pregunta}"),
])

chain = prompt | llm | StrOutputParser()
respuesta = chain.invoke({"contexto": contexto, "pregunta": pregunta})

print(f"Pregunta: {pregunta}")
print(f"Respuesta: {respuesta}")
````

## 10. Debugging: inspeccionar cada paso

Ventaja del pipeline RAG: podés revisar **cada paso por separado** para encontrar problemas.

````python
print("========== DEBUG RAG ==========")

q = "Que es LangChain?"
print(f"\n[1] Pregunta: {q}")

print(f"\n[2] Chunks recuperados:")
docs = retriever.invoke(q)
for i, d in enumerate(docs):
    print(f"  [{i+1}] {d.page_content}")

print(f"\n[3] Contexto armado:")
ctx = "\n\n".join(d.page_content for d in docs)
print(f"  {ctx}")

print(f"\n[4] Prompt formateado:")
msgs = prompt.format_messages(contexto=ctx, pregunta=q)
for m in msgs:
    print(f"  [{m.type}] {m.content[:120]}...")

print(f"\n[5] Respuesta:")
print(f"  {chain.invoke({'contexto': ctx, 'pregunta': q})}")
print("\n================================")
print("Si la respuesta es mala, revisa el paso donde falla.")
````

## 11. Errores comunes y cómo solucionarlos

| # | Error | Causa | Solución |
|---|---|---|---|
| 1 | El LLM inventa (alucina) | System prompt sin restricción de contexto | Agregá "Responde solo con el contexto dado. Si no está, decí que no sabés." |
| 2 | El retriever no encuentra nada | `chunk_size` muy grande/grande o `k` muy chico | Probá chunk_size=300-500, k=3 |
| 3 | Respuesta incompleta | `k` muy bajo (solo 1 doc) | Aumentá k a 3-5 |
| 4 | El loader falla | Encoding incorrecto | Usá `encoding="utf-8"` |
| 5 | Chroma da error | Falta instalar chromadb | `pip install chromadb` |
| 6 | Muchos tokens/gasto alto | `chunk_size` muy grande o `k` muy alto | Reducí chunk_size a 300-500, k a 2-3 |
| 7 | Contexto con información irrelevante | Chunks mal divididos (cortan frases) | Aumentá `chunk_overlap` |
| 8 | No se actualiza con datos nuevos | Chroma en memoria (no persistido) | Usá `Chroma(persist_directory=...)` |

### Checklist para debuggear un pipeline RAG

```text
[ ] El loader carga el documento correctamente?
[ ] Los chunks tienen sentido o cortan frases?
[ ] El retriever devuelve algo para la pregunta?
[ ] Los docs recuperados son relevantes?
[ ] El system prompt restringe al contexto?
[ ] La chain usa StrOutputParser?
[ ] El resultado es un string?
```

## 12. Comparación: sin RAG vs con RAG

| Aspecto | Sin RAG (LLM directo) | Con RAG |
|---|---|---|
| Conocimiento | Solo training data del LLM | Tus documentos + training data |
| Actualización | Esperar nuevo modelo | Cambiar docs en el vector store |
| Alucinaciones | Probables | Controladas por contexto |
| Costo por consulta | Bajo (solo prompt) | Medio (embedding + búsqueda + prompt) |
| Precisión en datos propios | Mala (no los conoce) | Alta |
| Transparencia | Caja negra | Sabés qué docs usó |
| Escala | Sin límite (pero sin datos tuyos) | Millones de documentos |

### Cuándo usar RAG ✅

- Tenés documentos propios (PDFs, manuales, bases de conocimiento)
- Necesitás información actualizada (no del training cutoff)
- Querés controlar las fuentes que usa el LLM
- Tenés más de 10 documentos que el LLM no conoce

### Cuándo NO usar RAG ❌

- Preguntas de cultura general (el LLM ya lo sabe)
- Tareas puramente creativas (no necesitan contexto)
- Prototipos de 1 consulta sin documentos propios

## Resumen final

### Pipeline completo de una línea

```text
Archivo -> TextLoader -> RecursiveCharacterTextSplitter -> OpenAIEmbeddings -> Chroma -> as_retriever -> invoke -> contexto -> ChatPromptTemplate -> ChatOpenAI -> StrOutputParser -> respuesta
```

### Conceptos clave

1. **Loader**: lleva el documento del mundo real a LangChain (`Document`)
2. **Splitter**: divide en fragmentos manejables (chunks)
3. **Embeddings**: convierte texto a vectores numéricos (1536 dimensiones)
4. **Vector Store**: guarda vectores + texto, permite búsqueda por similitud
5. **Retriever**: interfaz estándar para buscar los k fragmentos más relevantes
6. **Prompt**: arma el mensaje con contexto + pregunta + restricción de alucinaciones
7. **LLM**: genera la respuesta basada en el contexto
8. **Parser**: extrae el texto del objeto AIMessage
