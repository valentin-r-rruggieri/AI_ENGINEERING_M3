# M3L2 E02 - Chat con memoria (Resolution)

---
## 1. ¿Por qué un LLM no recuerda?

Los LLMs son **stateless** (sin estado): cada llamada es independiente. El modelo no guarda ninguna información entre llamadas.

```text
Llamada 1: usuario -> "Me llamo Ana"
  LLM responde: "Hola Ana!"
  -> El modelo NO guarda "Ana" en ningun lado

Llamada 2: usuario -> "Como me llamo?"
  LLM responde: ??  <- no lo sabe, cada llamada empieza de cero
```

### La solución: re-enviar el historial

"Memoria" en LLMs no es memoria real. Es simplemente **guardar los mensajes anteriores y re-enviarlos** en cada nueva llamada.

```text
Llamada 1 (con memoria):
  Input: [System + "Me llamo Ana"] -> LLM -> "Hola Ana!"
  -> Guardamos: usuario dijo "Me llamo Ana", AI dijo "Hola Ana!"

Llamada 2 (con memoria):
  Input: [System + ("Me llamo Ana", "Hola Ana!") + "Como me llamo?"]
  -> LLM ve el historial y responde: "Te llamas Ana"
```

### Conceptos clave

| Concepto | ¿Qué es? |
|---|---|
| **Contexto** | Información que se envía en la llamada actual |
| **Historial** | Mensajes anteriores que re-enviamos al modelo |
| **Memoria** | Mecanismo que guarda y recupera el historial automáticamente |
| **`session_id`** | Identificador único de una conversación (permite tener N conversaciones paralelas) |
| **`MessagesPlaceholder`** | Marcador en el prompt que indica DÓNDE inyectar el historial |
| **`RunnableWithMessageHistory`** | Wrapper que conecta la chain con el sistema de memoria |

### El flujo con memoria

```text
Pregunta del usuario
  |
  v
1. Buscar historial por session_id
  |
  v
2. Inyectar historial en el prompt (via MessagesPlaceholder)
  |
  v
3. Enviar prompt + historial + pregunta al LLM
  |
  v
4. LLM genera respuesta (viendo el historial)
  |
  v
5. Guardar (pregunta, respuesta) en el historial
  |
  v
6. Devolver respuesta al usuario
```

## 2. Setup: API key e imports

````python
import os, getpass
from langchain_openai import ChatOpenAI

if not os.getenv("OPENAI_API_KEY"):
    os.environ["OPENAI_API_KEY"] = getpass.getpass("OpenAI API key: ")

llm = ChatOpenAI(model="gpt-4o-mini", temperature=0)
print("API key + LLM listos")
````

## 3. Componentes de memoria

LangChain usa 4 piezas para armar memoria conversacional:

````python
from langchain_core.prompts import ChatPromptTemplate, MessagesPlaceholder
from langchain_core.output_parsers import StrOutputParser
from langchain_core.runnables.history import RunnableWithMessageHistory
from langchain_core.chat_history import InMemoryChatMessageHistory

parser = StrOutputParser()
historiales = {}  # diccionario: session_id -> historial
print("Componentes de memoria listos")
````

---
## 4. MessagesPlaceholder: el marcador de posición

**`MessagesPlaceholder("historial")`** es un marcador en el prompt que le dice a LangChain: *"acá van los mensajes del historial"*.

Sin esto, el prompt sería fijo y no podríamos inyectar dinámicamente los mensajes previos.

```python
prompt = ChatPromptTemplate.from_messages([
    ("system", "Sos un asistente..."),
    MessagesPlaceholder("historial"),   # <- aca se inyectan N mensajes
    ("human", "{pregunta}"),
])
```

Cuando se invoca la chain, `MessagesPlaceholder` se reemplaza por los mensajes que tenga el historial en ese momento.

````python
prompt = ChatPromptTemplate.from_messages([
    ("system", "Sos un asistente claro y breve. Usa el historial si ayuda a responder."),
    MessagesPlaceholder("historial"),
    ("human", "{pregunta}"),
])
cadena_base = prompt | llm | parser

print(f"Template creado con MessagesPlaceholder")
print(f"Variables del prompt: {prompt.input_variables}")
print(f"Donde va el historial: MessagesPlaceholder('historial')")
````

---
## 5. InMemoryChatMessageHistory: guardar mensajes

`InMemoryChatMessageHistory` es un almacén de mensajes en memoria (para producción se usa Redis, SQL, etc.).

Métodos principales:
- `add_user_message(texto)`: guarda un mensaje del usuario
- `add_ai_message(texto)`: guarda una respuesta del AI
- `get_messages()`: devuelve todos los mensajes guardados
- `clear()`: borra el historial

Cada conversación tiene su propio `InMemoryChatMessageHistory`, identificado por `session_id`.

### Función que obtiene/crea el historial por session_id

LangChain necesita una función que reciba un `session_id` y devuelva el historial correspondiente.

````python
def obtener_historial(session_id: str):
    if session_id not in historiales:
        historiales[session_id] = InMemoryChatMessageHistory()
    return historiales[session_id]

print(f"Funcion obtener_historial creada")
print(f"Historiales activos: {len(historiales)}")
````

---
## 6. RunnableWithMessageHistory: el wrapper de memoria

**`RunnableWithMessageHistory`** envuelve la chain y agrega el sistema de memoria automáticamente.

Parámetros:
- `runnable`: la chain base (prompt | llm | parser)
- `get_session_history`: función que devuelve el historial para un session_id
- `input_messages_key`: el nombre de la variable que contiene la pregunta del usuario ("pregunta")
- `history_messages_key`: el nombre del MessagesPlaceholder ("historial")

```text
Cuando invocas chat_con_memoria.invoke({"pregunta": "..."}, config={"session_id": "x"}):
  1. Llama a obtener_historial("x") -> obtiene InMemoryChatMessageHistory
  2. Toma los mensajes del historial y los pone en MessagesPlaceholder("historial")
  3. Ejecuta la chain: prompt | llm | parser
  4. Guarda automáticamente (pregunta, respuesta) en el historial
```

````python
chat_con_memoria = RunnableWithMessageHistory(
    cadena_base,
    obtener_historial,
    input_messages_key="pregunta",
    history_messages_key="historial",
)

print(f"Wrapper de memoria: {type(chat_con_memoria).__name__}")
print("Listo para invocar con session_id")
````

---
## 7. Probar la memoria

Invocamos con **el mismo `session_id`** para que el modelo recuerde la conversación anterior.

````python
config = {"configurable": {"session_id": "alumno-1"}}

respuesta_1 = chat_con_memoria.invoke(
    {"pregunta": "Me llamo Ana y estoy aprendiendo LangChain."},
    config=config
)
respuesta_2 = chat_con_memoria.invoke(
    {"pregunta": "Como me llamo y que estoy aprendiendo?"},
    config=config
)

print("Respuesta 1 (presentacion):")
print(f"  {respuesta_1}\n")
print("Respuesta 2 (con memoria):")
print(f"  {respuesta_2}")
````

### Verificar que realmente hay mensajes guardados

````python
for sid, hist in historiales.items():
    print(f"=== Historial: {sid} ===")
    for msg in hist.messages:
        print(f"  [{msg.type}] {msg.content}")
    print()
print(f"Total de historiales: {len(historiales)}")
````

### Probar con otro session_id (memoria separada)

Cada `session_id` tiene su propio historial independiente.

````python
config_b = {"configurable": {"session_id": "alumno-2"}}

r_b = chat_con_memoria.invoke(
    {"pregunta": "Como me llamo y que estoy aprendiendo?"},
    config=config_b
)

print(f"session_id='alumno-2' (sin historial previo):")
print(f"  {r_b}")
print()
print("'alumno-2' no tiene historial, el modelo no sabe quien es.")
print("'alumno-1' si tiene historial, el modelo sabe que se llama Ana.")
````

## 8. Debugging: inspeccionar el historial en cada paso

````python
session = "debug-sesion"
cfg = {"configurable": {"session_id": session}}

print("========== DEBUG MEMORIA ==========")

print("\n[Antes de invocar] Historial vacio:")
h = obtener_historial(session)
print(f"  Mensajes: {len(h.messages)}")

print("\n[Invocacion 1] 'Hola, soy Pedro'")
r1 = chat_con_memoria.invoke({"pregunta": "Hola, soy Pedro"}, config=cfg)
print(f"  Respuesta: {r1}")
print(f"  Mensajes en historial: {len(h.messages)}")
for m in h.messages:
    print(f"    [{m.type}] {m.content}")

print("\n[Invocacion 2] 'Como me llamo?'")
r2 = chat_con_memoria.invoke({"pregunta": "Como me llamo?"}, config=cfg)
print(f"  Respuesta: {r2}")
print(f"  Mensajes en historial: {len(h.messages)}")
for m in h.messages:
    print(f"    [{m.type}] {m.content}")
print("\n==================================")
````

## 9. Errores comunes

| # | Error | Causa | Solución |
|---|---|---|---|
| 1 | El modelo no recuerda | Se usa distinto `session_id` en cada invocación | Usá el MISMO `session_id` |
| 2 | "Variable `historial` not found" | El nombre del `MessagesPlaceholder` no coincide con `history_messages_key` | Verificá que ambos se llamen igual |
| 3 | Todos los usuarios comparten memoria | Se usa siempre el mismo `session_id` | Usá un `session_id` distinto por usuario |
| 4 | El historial crece infinito | No hay límite de mensajes guardados | Implementá poda (ej: últimos 10 mensajes) |
| 5 | Error en `config` | El dict de configuración está mal formado | Usá `{"configurable": {"session_id": "..."}}` |
| 6 | Se duplica el historial | `RunnableWithMessageHistory` agregado dos veces | Asegurate de envolver la chain una sola vez |

### Límites de la memoria en memoria

| Problema | Consecuencia | Solución en producción |
|---|---|---|
| `InMemoryChatMessageHistory` | Se pierde al reiniciar el proceso | Usá `RedisChatMessageHistory` o `SQLChatMessageHistory` |
| Tokens ilimitados | El historial puede exceder el contexto del modelo | Limitá a los últimos N mensajes o N tokens |
| Costo creciente | Más historial = más tokens por llamada | Resumí el historial antiguo (ej: cada 10 interacciones) |

---
## 10. Comparación: sin memoria vs con memoria

| Aspecto | Sin memoria | Con memoria (`RunnableWithMessageHistory`) |
|---|---|---|
| La llamada anterior... | Se pierde | Se guarda y re-envía |
| Preguntas de seguimiento | No funciona | Funciona |
| Identidad del usuario | No la recuerda | La recuerda por `session_id` |
| Conversaciones paralelas | No aplica | Sí, cada `session_id` es independiente |
| Código extra | Ninguno | `MessagesPlaceholder` + `RunnableWithMessageHistory` + función get_history |
| Costo por llamada | Solo el prompt actual | Prompt actual + historial (más tokens) |

### ¿Cuándo usar memoria?

- **Siempre** que el usuario haga preguntas de seguimiento
- Cuando necesitás mantener contexto entre turnos
- Asistentes conversacionales con identidad de usuario

### ¿Cuándo NO?

- Operaciones one-shot (una pregunta, una respuesta)
- Cuando el costo de tokens extra no se justifica
- Tareas puramente transaccionales (ej: clasificar un texto)

## Resumen final

### El concepto central

> **La memoria en LLMs no es magia. Es historial guardado y re-enviado al modelo en cada llamada.**

### Lo que aprendiste

1. Los LLMs son **stateless**: cada llamada empieza de cero
2. **MessagesPlaceholder** marca dónde va el historial en el prompt
3. **InMemoryChatMessageHistory** guarda los mensajes en memoria
4. **RunnableWithMessageHistory** conecta la chain con el sistema de memoria
5. **session_id** identifica conversaciones de forma única
6. Cada `session_id` tiene su propio historial independiente
